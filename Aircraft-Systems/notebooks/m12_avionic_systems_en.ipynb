## 1. The GPS orbit
Does the orbital period at 20,200 km agree with Tooley's 12 hours, and what is 20,200 km in nautical miles?

```python
import math
# Quy dao GPS: Kepler (hang so chuan) va ti le theo quy dao Mat Trang (khong dung mu)
R, mu = 6371e3, 3.986004418e14
r = R + 20200e3
T1 = 2 * math.pi * math.sqrt(r ** 3 / mu) / 3600
T2 = 27.321661 * 24 * (r / 384400e3) ** 1.5
print('Kepler:', round(T1, 3), 'h; ti le Mat Trang:', round(T2, 3), 'h')
print('20200 km =', round(20200 / 1.852), 'hai ly (Tooley lam tron 10900)')
```

**Reading the result.** Kepler gives 11.973 hours and scaling from the Moon gives 11.917 hours: both are close to 12. 20,200 km is 10907 nautical miles (the book says 10,900).

## 2. GPS codes and the FMS cycle
How many chips in one C/A period, and how many FMS updates a year?

```python
# Ma C/A va P, chu ky cap nhat FMS
print('C/A: so chip moi chu ky =', round(1.023e6 * 1e-3))
print('P: 267 ngay / 7 ngay =', 267 // 7, 'doan day du')
print('FMS: chu ky moi nam =', 365 // 28, '; 13 x 28 =', 13 * 28, 'ngay')
```

**Reading the result.** The C/A code has 1,023 chips. The P code splits into 38 full seven-day segments. The FMS has 13 cycles a year, 364 days in all.

## 3. The IRS integrating acceleration
The velocity and distance after 30 s at 2 m/s², and the error when the accelerometer is off by 0.01 m/s².

```python
# IRS: tich phan gia toc khong doi; sai so cua gia toc ke
def integrate(a, t, n=100000):
    dt = t / n; v = s = 0.0
    for _ in range(n): s += v * dt + 0.5 * a * dt * dt; v += a * dt
    return v, s
print('a = 2, t = 30:', [round(x, 3) for x in integrate(2, 30)], '| cong thuc:', 2 * 30, 0.5 * 2 * 30 ** 2)
print('sai so 0,01 m/s2 sau 40 s:', [round(x, 3) for x in integrate(0.01, 40)])
```

**Reading the result.** Numerical summation gives 60 m/s and 900 m, matching the formulas. An error of 0.01 m/s² becomes 0.4 m/s and 8 m after 40 s.

## 4. TCAS
The time τ for a few (distance, closure rate) pairs.

```python
# TCAS: thoi gian toi diem tiep can gan nhat tau = d / v_dong
for d, v in ((10, 600), (4, 480), (6, 360), (8, 240 + 240)):
    print(d, 'NM,', v, 'kt ->', round(d / v * 3600), 's')
```

**Reading the result.** 10 NM at 600 kt is 60 s; 4 NM at 480 kt is 30 s; two 240 kt aircraft head-on 8 NM apart close at 480 kt and τ = 60 s.
