## 1. Base converter
🎯 **What question does this method answer?** Given a number in any base (2/8/10/16), how to convert it exactly into another base without computing by hand, and how to reuse the converter to check many questions in bulk.

```python
def to_decimal(s: str, base: int) -> int:
    return int(s, base)

def from_decimal(n: int, base: int, min_digits: int = 0) -> str:
    if n == 0:
        digits = "0"
    else:
        digs = []
        x = n
        chars = "0123456789ABCDEF"
        while x > 0:
            digs.append(chars[x % base])
            x //= base
        digits = "".join(reversed(digs))
    return digits.zfill(min_digits)

def decimal_to_bcd(n: int) -> str:
    return "".join(format(int(c), "04b") for c in str(n))

def bcd_to_decimal(bcd: str) -> int:
    assert len(bcd) % 4 == 0, "BCD length must be a multiple of 4 bits"
    groups = [bcd[i:i+4] for i in range(0, len(bcd), 4)]
    return int("".join(str(int(g, 2)) for g in groups))

def twos_complement(bits: str) -> str:
    inv = "".join("1" if b == "0" else "0" for b in bits)
    return from_decimal(int(inv, 2) + 1, 2, len(bits))

# quick demo
print("21 -> binary:", from_decimal(21, 2))
print("10101(2) -> decimal:", to_decimal("10101", 2))
print("BCD of 37:", decimal_to_bcd(37))
print("Two's complement of 10110:", twos_complement("10110"))
```

#### 📤 Real output
The functions above reproduce the principles taught: base conversion by repeated division/multiplication, BCD split into groups of 4 bits, two's complement = invert the bits + add 1. The next part uses these functions to check all 13 original questions.

## 2. Re-checking the 13 original questions (Tooley, ch.2) with code
🎯 **What question does this method answer?** Are the answers displayed on the website (`data/quiz/01-number-systems.*.json`) really correct: independent verification by automatic computation instead of trusting hand calculation.

```python
checks = []

checks.append(("10101(2) = ? (10)", to_decimal("10101", 2), 21))
checks.append(("29(10) = ? (2)", from_decimal(29, 2), "11101"))
checks.append(("Two's complement of 10110", twos_complement("10110"), "01010"))
checks.append(("BCD 10010001 = ? (10)", bcd_to_decimal("10010001"), 91))
checks.append(("37(10) = ? BCD", decimal_to_bcd(37), "00110111"))
checks.append(("73(8) = ? (10)", to_decimal("73", 8), 59))
checks.append(("111(16) = ? (10) roi -> (8)", from_decimal(to_decimal("111", 16), 8), "421"))
checks.append(("C9(16) = ? (10)", to_decimal("C9", 16), 201))
checks.append(("10110011(2) = ? (16)", from_decimal(to_decimal("10110011", 2), 16), "B3"))
checks.append(("AD(16) = ? (2, 8-bit)", from_decimal(to_decimal("AD", 16), 2, 8), "10101101"))
checks.append(("706(8) = ? (10) roi -> (16)", from_decimal(to_decimal("706", 8), 16), "1C6"))

ok_count = 0
for label, got, expect in checks:
    ok = str(got).upper() == str(expect).upper()
    ok_count += ok
    print(f"{'OK ' if ok else 'BAD'}  {label:32s} -> computed={got!s:12s} expected={expect}")

print(f"\nTotal: {ok_count}/{len(checks)} questions match the answers published on the website.")
```
