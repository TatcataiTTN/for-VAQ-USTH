## 1. Toán tử và độ ưu tiên
Năm biểu thức của Tooley 17.7 với p = 1010 và q = 0110.

```python
# Toan tu VHDL theo thu tu uu tien (NOT > dich/xoay > AND/OR/XOR) tren vector 4 bit
M = 0xF
nt = lambda v: ~v & M
sla = lambda v: (v << 1) & M
ror = lambda v: ((v >> 1) | ((v & 1) << 3)) & M
p, q = 0b1010, 0b0110
for name, val in (('p AND q', p & q), ('NOT p AND q', nt(p) & q), ('p AND NOT q', p & nt(q)), ('p AND SLA q', p & sla(q)), ('NOT p AND ROR q', nt(p) & ror(q))):
    print(f'{name:18s}', format(val, '04b'))
print('NOT (p AND q) =', format(nt(p & q), '04b'), '(dau ngoac doi thu tu)')
```

**Đọc kết quả.** Kết quả 0010, 0100, 1000, 1000, 0001 khớp sách. NOT (p AND q) = 1101 khác NOT p AND q = 0100 vì dấu ngoặc đổi thứ tự.

## 2. Bộ nửa cộng
Bảng chân trị của bộ nửa cộng và bản có EN.

```python
# Bo nua cong: bang chan tri hanh vi (XOR, AND) so voi phep cong so hoc; va ban co EN
for x in (0, 1):
    for y in (0, 1):
        s, c = x ^ y, x & y
        assert (s, c) == ((x + y) % 2, (x + y) // 2)
        print(x, y, '->', s, c, '| EN = 0:', 0, 0)
```

**Đọc kết quả.** Sum là XOR, carry là AND; trùng phép cộng số học. Khi EN = 0 cả hai ngõ ra bằng 0.

## 3. Trễ lan truyền
Trễ tổng của vài đường tín hiệu và thời điểm c đổi.

```python
# Tre lan truyen cong don
for n, t in ((3, 15), (4, 15), (5, 12)): print(n, 'cong x', t, 'ns =', n * t, 'ns')
print('XOR 15 ns: a doi tai 100 ns -> c doi tai', 100 + 15, 'ns')
```

**Đọc kết quả.** Ba cổng 15 ns trễ 45 ns (như bộ nửa cộng của Tooley); c đổi tại 115 ns khi a đổi tại 100 ns.
