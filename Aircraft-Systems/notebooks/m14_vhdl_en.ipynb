## 1. Operators and precedence
Tooley's five expressions of 17.7 with p = 1010 and q = 0110.

```python
# Toan tu VHDL theo thu tu uu tien (NOT > dich/xoay > AND/OR/XOR) tren vector 4 bit
M = 0xF
nt = lambda v: ~v & M
sla = lambda v: (v << 1) & M
ror = lambda v: ((v >> 1) | ((v & 1) << 3)) & M
p, q = 0b1010, 0b0110
for name, val in (('p AND q', p & q), ('NOT p AND q', nt(p) & q), ('p AND NOT q', p & nt(q)), ('p AND SLA q', p & sla(q)), ('NOT p AND ROR q', nt(p) & ror(q))):
    print(f'{name:18s}', format(val, '04b'))
print('NOT (p AND q) =', format(nt(p & q), '04b'), '(dau ngoac doi thu tu)')
```

**Reading the result.** The results 0010, 0100, 1000, 1000, 0001 match the book. NOT (p AND q) = 1101 differs from NOT p AND q = 0100 because the brackets change the order.

## 2. The half adder
The truth table of the half adder and the version with EN.

```python
# Bo nua cong: bang chan tri hanh vi (XOR, AND) so voi phep cong so hoc; va ban co EN
for x in (0, 1):
    for y in (0, 1):
        s, c = x ^ y, x & y
        assert (s, c) == ((x + y) % 2, (x + y) // 2)
        print(x, y, '->', s, c, '| EN = 0:', 0, 0)
```

**Reading the result.** Sum is the XOR and carry the AND, equal to arithmetic addition. With EN = 0 both outputs are 0.

## 3. Propagation delay
The total delay of a few signal paths and when c changes.

```python
# Tre lan truyen cong don
for n, t in ((3, 15), (4, 15), (5, 12)): print(n, 'cong x', t, 'ns =', n * t, 'ns')
print('XOR 15 ns: a doi tai 100 ns -> c doi tai', 100 + 15, 'ns')
```

**Reading the result.** Three 15 ns gates delay 45 ns (like Tooley's half adder); c changes at 115 ns when a changes at 100 ns.
