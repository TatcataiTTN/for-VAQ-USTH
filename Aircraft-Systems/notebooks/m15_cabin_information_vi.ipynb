## 1. Chuông cabin
Chu kỳ, số chu kỳ trong 900 ms và tỉ số tần số của hai chuông.

```python
# Chuong cabin: 494 Hz va 587 Hz keo dai 900 ms
for f in (494, 587):
    print(f, 'Hz: chu ky', round(1000 / f, 3), 'ms; so chu ky trong 900 ms =', round(f * 0.9, 1))
print('ti so tan so 587/494 =', round(587 / 494, 3), '(gan 6/5 = 1.2, khong phai quang tam = 2)')
```

**Đọc kết quả.** 494 Hz có chu kỳ 2,024 ms và 444,6 chu kỳ; 587 Hz có chu kỳ 1,704 ms và 528,3 chu kỳ; tỉ số 1,188 gần một quãng ba thứ.

## 2. Nạp phương tiện IFE
Thời gian nạp tệp 4,7 GB và 20 GB qua 100 Mbit/s và 1 Gbit/s.

```python
# Thoi gian nap tep phuong tien vao bo nho IFE (gia dinh cua site: 1 GB = 1e9 byte, bo qua mao dau)
for size in (4.7, 20):
    for name, rate in (('100 Mbit/s', 100e6), ('1 Gbit/s', 1e9)):
        t = size * 8e9 / rate; print(size, 'GB qua', name, '->', round(t, 1), 's =', round(t / 60, 2), 'phut')
```

**Đọc kết quả.** 4,7 GB mất 376 s (6,3 phút) ở 100 Mbit/s và 37,6 s ở 1 Gbit/s; tệp 20 GB mất 1 600 s và 160 s. Gigabit nhanh gấp mười lần.

## 3. Tốc độ cất cánh
Thứ tự và chênh lệch của V1, VR, V2.

```python
# Toc do cat canh B737 dien hinh (Tooley 20.1)
v = {'V1': 124, 'VR': 125, 'V2': 136}
print(sorted(v, key=v.get), '; V2 - V1 =', v['V2'] - v['V1'], 'kt')
```

**Đọc kết quả.** Thứ tự V1, VR, V2 (124, 125, 136 kt); V2 hơn V1 là 12 kt.
