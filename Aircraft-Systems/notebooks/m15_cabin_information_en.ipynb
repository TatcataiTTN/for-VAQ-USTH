## 1. The cabin chimes
The period, the number of cycles in 900 ms and the frequency ratio of the two chimes.

```python
# Chuong cabin: 494 Hz va 587 Hz keo dai 900 ms
for f in (494, 587):
    print(f, 'Hz: chu ky', round(1000 / f, 3), 'ms; so chu ky trong 900 ms =', round(f * 0.9, 1))
print('ti so tan so 587/494 =', round(587 / 494, 3), '(gan 6/5 = 1.2, khong phai quang tam = 2)')
```

**Reading the result.** 494 Hz has a 2.024 ms period and 444.6 cycles; 587 Hz has a 1.704 ms period and 528.3 cycles; the ratio 1.188 is close to a minor third.

## 2. IFE media loading
The loading time of a 4.7 GB and a 20 GB file over 100 Mbit/s and 1 Gbit/s.

```python
# Thoi gian nap tep phuong tien vao bo nho IFE (gia dinh cua site: 1 GB = 1e9 byte, bo qua mao dau)
for size in (4.7, 20):
    for name, rate in (('100 Mbit/s', 100e6), ('1 Gbit/s', 1e9)):
        t = size * 8e9 / rate; print(size, 'GB qua', name, '->', round(t, 1), 's =', round(t / 60, 2), 'phut')
```

**Reading the result.** A 4.7 GB file takes 376 s (6.3 min) at 100 Mbit/s and 37.6 s at 1 Gbit/s; a 20 GB file takes 1,600 s and 160 s. Gigabit is ten times faster.

## 3. Take-off speeds
The order and difference of V1, VR, V2.

```python
# Toc do cat canh B737 dien hinh (Tooley 20.1)
v = {'V1': 124, 'VR': 125, 'V2': 136}
print(sorted(v, key=v.get), '; V2 - V1 =', v['V2'] - v['V1'], 'kt')
```

**Reading the result.** The order is V1, VR, V2 (124, 125, 136 kt); V2 exceeds V1 by 12 kt.
