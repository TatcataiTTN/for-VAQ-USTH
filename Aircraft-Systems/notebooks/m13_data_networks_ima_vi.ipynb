## 1. BAG
Bảng 16.1 và BAG lớn nhất cho 100 và 400 thông điệp mỗi giây.

```python
# Bang 16.1 va chon BAG
BAGS = [1, 2, 4, 8, 16, 32, 64, 128]
print({b: 1000 / b for b in BAGS})
def max_bag(rate): return max(b for b in BAGS if 1000 / b >= rate)
print('100/s ->', max_bag(100), 'ms; 400/s ->', max_bag(400), 'ms (khoa in cua sach: 4 ms)')
```

**Đọc kết quả.** BAG 1 đến 128 ms ứng với 1 000 đến 7,8125 Hz. 100 lần mỗi giây cho BAG 8 ms (khớp ví dụ của Tooley); 400 lần mỗi giây cho 2 ms, vì 4 ms chỉ cho 250 Hz.

## 2. Tốc độ bit của VL
Tốc độ bit lớn nhất cho khung 1 518 byte với BAG 4 ms và khung 256 byte với BAG 128 ms.

```python
# Toc do bit lon nhat cua VL = L * 8 * 1000 / BAG
for L, bag in ((1518, 4), (256, 128), (64, 2)): print(L, 'byte,', bag, 'ms ->', L * 8 * 1000 / bag, 'bit/s')
```

**Đọc kết quả.** 3 036 000 bit/s (≈ 3 Mbit/s, đúng ví dụ Tooley) và 16 000 bit/s (câu 6 Chương 16).

## 3. Khung AFDX
Độ dài nhỏ nhất, lớn nhất và thời gian truyền.

```python
# Do dai khung AFDX tu cac truong
def frame_len(payload):
    pad = max(0, 17 - payload)
    return 6 + 6 + 2 + 20 + 8 + payload + pad + 1 + 4
print('payload 1:', frame_len(1), '| payload 17:', frame_len(17), '| payload 1471:', frame_len(1471))
for nbytes, rate in ((1518, 10e6), (1518, 100e6)): print(nbytes, 'byte o', rate / 1e6, 'Mbit/s:', nbytes * 8 / rate * 1e3, 'ms')
print('khoang cach giua khung 12 byte o 100 Mbit/s:', 12 * 8 / 100e6 * 1e6, 'us')
```

**Đọc kết quả.** Payload 1 (đệm 16) và payload 17 đều cho 64 byte; payload 1 471 cho 1 518 byte. Khung dài nhất mất 1,2144 ms ở 10 Mbit/s và 0,12144 ms ở 100 Mbit/s; khoảng cách 12 byte là 0,96 μs ở 100 Mbit/s.

## 4. Số thứ tự và lùi hàm mũ
SN sau 255 là gì, và khoảng lùi sau 1 đến 5 va chạm?

```python
# So thu tu quay vong va lui ham mu nhi phan
sn, seq = 253, []
for _ in range(5): seq.append(sn); sn = 1 if sn == 255 else sn + 1
print(seq)
print([2 ** (n - 1) for n in range(1, 6)], '(boi cua T sau lan va cham thu n)')
```

**Đọc kết quả.** Sau 255 là 1 (không phải 0). Khoảng lùi là 1T, 2T, 4T, 8T, 16T.
