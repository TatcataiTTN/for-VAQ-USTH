## 1. Quỹ đạo GPS
Chu kỳ quỹ đạo ở độ cao 20 200 km có khớp 12 giờ của Tooley không, và đổi 20 200 km sang hải lý?

```python
import math
# Quy dao GPS: Kepler (hang so chuan) va ti le theo quy dao Mat Trang (khong dung mu)
R, mu = 6371e3, 3.986004418e14
r = R + 20200e3
T1 = 2 * math.pi * math.sqrt(r ** 3 / mu) / 3600
T2 = 27.321661 * 24 * (r / 384400e3) ** 1.5
print('Kepler:', round(T1, 3), 'h; ti le Mat Trang:', round(T2, 3), 'h')
print('20200 km =', round(20200 / 1.852), 'hai ly (Tooley lam tron 10900)')
```

**Đọc kết quả.** Kepler cho 11,973 giờ, tỉ lệ theo Mặt Trăng cho 11,917 giờ: cả hai gần 12 giờ. 20 200 km là 10907 hải lý (sách ghi 10 900).

Một chu kỳ mã C/A có bao nhiêu chip và FMS cập nhật mấy lần mỗi năm?

## 3. IRS tích phân gia tốc
Vận tốc và quãng đường sau 30 s với 2 m/s², và sai số khi gia tốc kế lệch 0,01 m/s².

```python
# IRS: tich phan gia toc khong doi; sai so cua gia toc ke
def integrate(a, t, n=100000):
    dt = t / n; v = s = 0.0
    for _ in range(n): s += v * dt + 0.5 * a * dt * dt; v += a * dt
    return v, s
print('a = 2, t = 30:', [round(x, 3) for x in integrate(2, 30)], '| cong thuc:', 2 * 30, 0.5 * 2 * 30 ** 2)
print('sai so 0,01 m/s2 sau 40 s:', [round(x, 3) for x in integrate(0.01, 40)])
```

**Đọc kết quả.** Phép lấy tổng số cho 60 m/s và 900 m, khớp công thức. Sai số 0,01 m/s² sau 40 s thành 0,4 m/s và 8 m.

## 4. TCAS
Thời gian τ cho vài cặp (khoảng cách, tốc độ tiếp cận).

```python
# TCAS: thoi gian toi diem tiep can gan nhat tau = d / v_dong
for d, v in ((10, 600), (4, 480), (6, 360), (8, 240 + 240)):
    print(d, 'NM,', v, 'kt ->', round(d / v * 3600), 's')
```

**Đọc kết quả.** 10 NM ở 600 kt là 60 s; 4 NM ở 480 kt là 30 s; hai máy bay 240 kt ngược chiều cách 8 NM có tốc độ tiếp cận 480 kt và τ = 60 s.
